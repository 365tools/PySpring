# Debug Service Naming Generation

This notebook checks how `dependency_injector` or the custom `AppContainerManager.generate_name` handles class names like `DBManagerService`.

```python
import re


def generate_name(service_class_name: str):
    return re.sub(r'(?<!^)(?=[A-Z])', '_', service_class_name).lower()


print(f"DBManagerService -> {generate_name('DBManagerService')}")
print(f"LogManager -> {generate_name('LogManager')}")
print(f"RepositoriesConfigManager -> {generate_name('RepositoriesConfigManager')}")
```
